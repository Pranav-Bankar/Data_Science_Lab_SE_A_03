```
prime_numbers = []

for number in range(2, 101):
    is_prime = True
    for i in range(2, int(number**0.5) + 1):
        if number % i == 0:
            is_prime = False
            break
    if is_prime:
        prime_numbers.append(number)

print(f"Prime numbers up to 100: {prime_numbers}")
```

## Simulate Coin Toss Experiment

### Subtask:
Simulate a coin toss experiment for a specified number of trials and compute the probabilities of getting heads and tails.

```
import random

# Number of coin toss trials
num_trials = 1000

# Initialize counters for heads and tails
heads_count = 0
tails_count = 0

# Simulate coin tosses
for _ in range(num_trials):
    # random.choice(['Heads', 'Tails']) simulates a fair coin toss
    outcome = random.choice(['Heads', 'Tails'])
    if outcome == 'Heads':
        heads_count += 1
    else:
        tails_count += 1

# Compute probabilities
probability_heads = heads_count / num_trials
probability_tails = tails_count / num_trials

# Print the results
print(f"Coin Toss Experiment Results ({num_trials} trials):")
print(f"Heads: {heads_count} (Probability: {probability_heads:.2f})")
print(f"Tails: {tails_count} (Probability: {probability_tails:.2f})")
```

### Subtask:
Generate a bar chart to visualize the results of the coin toss experiment.

```
import matplotlib.pyplot as plt
import numpy as np

# Data from the coin toss experiment
labels = ['Heads', 'Tails']
counts = [heads_count, tails_count]
probabilities = [probability_heads, probability_tails]

x = np.arange(len(labels)) # the label locations
width = 0.35 # the width of the bars

fig, ax1 = plt.subplots(figsize=(8, 6))

# Plot counts on the primary y-axis
rects1 = ax1.bar(x - width/2, counts, width, label='Counts', color='skyblue')
ax1.set_ylabel('Number of Occurrences', color='skyblue')
ax1.tick_params(axis='y', labelcolor='skyblue')

# Create a second y-axis for probabilities
ax2 = ax1.twinx()
rects2 = ax2.bar(x + width/2, probabilities, width, label='Probabilities', color='lightcoral')
ax2.set_ylabel('Probability', color='lightcoral')
ax2.tick_params(axis='y', labelcolor='lightcoral')

# Add some text for labels, title and custom x-axis tick labels, etc.
ax1.set_xlabel('Outcome')
ax1.set_title(f'Coin Toss Experiment Results ({num_trials} trials)')
ax1.set_xticks(x)
ax1.set_xticklabels(labels)

fig.tight_layout() # otherwise the right y-label is slightly clipped
plt.show()
```
